```python
import os

podfoldery = ["osobowe", "ciezarowe", "przemyslowe"]
samochody = ["maluch", "polonez", "syrena"]

for zmienna in podfoldery:
    for zmienna2 in samochody:
        os.makedirs(f"Samochody/{zmienna}/{zmienna2}", exist_ok=True)
os.chmod("Samochody", 0o750)
```

```python
import re

tekst = "'a' i 'b'"
positions = [i for i, c in enumerate(tekst) if c == "'"]

for i in range(len(positions)):
    for j in range(i + 1, len(positions)):
        start = positions[i]
        koniec = positions[j]
        podciag_z_znakami = tekst[start:koniec + 1]     
        if len(podciag_z_znakami) >= 2:
            wew = podciag_z_znakami[1:-1]
        else:
            wew = podciag_z_znakami    
        oddziel = re.sub(r"\s+", "", wew)
        print(oddziel)
```
